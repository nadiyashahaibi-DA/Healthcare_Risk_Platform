## 2. Why a generic parser is needed

Every PORDATA "Quadro" export follows the same quirky layout, and it is **not** a clean rectangular
table — `pd.read_excel(..., header=0)` alone will not produce a usable DataFrame:

- The real header sits several blank rows down (row index varies per file).
- Row `Territórios` (2 rows above the real header) carries the **indicator name(s)** as column labels.
- Row `Âmbito Geográfico | Anos | <values...>` is the real header — confusingly, the column literally
  labelled `Anos` ("Years") does **not** contain a year, it contains the **territory name**; the year
  itself only shows up as the data column's own header (e.g. `2025`).
- Below the real data there is a **legend/footnote block** (symbols like `┴`, `x`, `fr`, `Confidencial`,
  `Ausência de valor`, `Dado de fiabilidade reduzida`, ...) sitting in the *same two columns* as the data,
  which will corrupt naive `value_counts()` / groupby operations if not filtered out.
- To the right of the real ~3–6 data columns, the sheet carries a **wide, mostly-empty block of up to
  ~250 columns** — a leftover territory dropdown/validation list from the PORDATA export UI, not data.

The function below isolates just the real table: `Âmbito Geográfico == "Município"` rows, the correct
number of indicator columns (read off the `Territórios` row), nothing else.


```python
def load_pordata_quadro(path: Path, level: str = "Município") -> pd.DataFrame:
    """Parse a PORDATA 'Quadro' sheet export down to a tidy table at the requested
    geographic level (default: município). Returns columns ['Territorio', <indicator(s)>].
    Raises ValueError if the sheet has no real data rows (e.g. an empty/broken export).
    """
    raw = pd.read_excel(path, sheet_name="Quadro", header=None)

    hits = raw.index[raw[0] == "Âmbito Geográfico"]
    if len(hits) == 0:
        raise ValueError(f"{path.name}: no 'Âmbito Geográfico' header found — file appears to contain "
                          f"no data table (likely an empty/broken PORDATA export).")
    hdr_idx = hits[0]
    field_idx = hdr_idx - 2  # 'Territórios' row holding the indicator name(s)
    field_row = raw.iloc[field_idx]

    n_fields, c = 0, 2
    while c < raw.shape[1] and pd.notna(field_row[c]):
        n_fields += 1
        c += 1
    if n_fields == 0:
        raise ValueError(f"{path.name}: header row found but no indicator columns detected next to it.")

    field_names = [field_row[2 + i] for i in range(n_fields)]
    ncols = 2 + n_fields
    body = raw.iloc[hdr_idx + 1:, :ncols].copy()
    body.columns = ["Ambito", "Territorio"] + field_names

    out = body[body["Ambito"] == level].drop(columns="Ambito").reset_index(drop=True)
    for fn in field_names:
        out[fn] = pd.to_numeric(out[fn], errors="coerce")
    return out, field_names
```

## 3. Load and quality-check each raw file


### 3.1 `Distribution_population.xlsx` — population density


```python
density_df, density_fields = load_pordata_quadro(RAW_DIR / "Distribution_population.xlsx")
density_df = density_df.rename(columns={density_fields[0]: "densidade_hab_km2"})
print(f"rows: {len(density_df)}  |  missing values: {density_df['densidade_hab_km2'].isna().sum()}  "
      f"|  duplicate municípios: {density_df['Territorio'].duplicated().sum()}")
density_df.head()
```

### 3.2 `pharmacies_municipality.xlsx` — pharmacy counts


```python
pharm_df, pharm_fields = load_pordata_quadro(RAW_DIR / "pharmacies_municipality.xlsx")
pharm_df = pharm_df.rename(columns={pharm_fields[0]: "farmacias"})
print(f"rows: {len(pharm_df)}  |  missing values: {pharm_df['farmacias'].isna().sum()}  "
      f"|  duplicate municípios: {pharm_df['Territorio'].duplicated().sum()}")
pharm_df.head()
```

### 3.3 `professionals_municipality.xlsx` — médicos / enfermeiros / farmacêuticos / dentistas counts


```python
prof_df, prof_fields = load_pordata_quadro(RAW_DIR / "professionals_municipality.xlsx")
print("indicator columns found:", prof_fields)
print(f"rows: {len(prof_df)}  |  duplicate municípios: {prof_df['Territorio'].duplicated().sum()}")
print("missing values per column:")
print(prof_df[prof_fields].isna().sum())
prof_df.head()
```

```python
# Reconciliation check: do the 308 município-level counts sum to PORDATA's own national total row?
raw_prof = pd.read_excel(RAW_DIR / "professionals_municipality.xlsx", sheet_name="Quadro", header=None)
hdr_idx = raw_prof.index[raw_prof[0] == "Âmbito Geográfico"][0]
pt_row = raw_prof.iloc[hdr_idx + 1:]
pt_row = pt_row[(pt_row[0] == "NUTS 2024") & (pt_row[1] == "Portugal")]
national_totals = pd.to_numeric(pt_row.iloc[0, 2:2 + len(prof_fields)]).values

recon = pd.DataFrame({
    "indicator": prof_fields,
    "sum_of_308_municipios": [prof_df[f].sum() for f in prof_fields],
    "portugal_national_total": national_totals,
})
recon["diff"] = recon["sum_of_308_municipios"] - recon["portugal_national_total"]
recon["diff_pct"] = (recon["diff"] / recon["portugal_national_total"] * 100).round(2)
recon
```

**Reading this check:** Médicos, Enfermeiros and Dentistas reconcile exactly (0 diff). **Farmacêuticos**
does not — the 308 municípios sum to ~485 fewer than PORDATA's national total (~2.8% gap). This is a
genuine PORDATA data-quality quirk (INE's own metadata for this file notes pharmacists are geo-tagged by
*establishment location*, and a small number are evidently not attributed to a município in the source
register) — not a bug in this parser. **Keep this in mind if Farmacêuticos rates are used in the final
deliverable**: municipal Farmacêuticos figures will slightly under-count relative to the national picture.


### 3.4 `pharmacies_perthousand.xlsx` — national-level provider rates (not by município)


```python
raw_rates = pd.read_excel(RAW_DIR / "pharmacies_perthousand.xlsx", sheet_name="Quadro", header=None)
hdr_idx = raw_rates.index[raw_rates[0] == "Anos"][0]
rate_cols = raw_rates.iloc[hdr_idx]
year_row = raw_rates.iloc[hdr_idx + 1]
national_rates = pd.Series(year_row.values[1:6], index=rate_cols.values[1:6], name="per_100k_hab_2025")
national_rates = pd.to_numeric(national_rates)
national_rates.to_frame()
```

This file has **one row**: Portugal's national rate per 100,000 residents for 2025, for five professions.
Note **Odontologistas = 0** — PORDATA's 2025 methodology appears to have folded that category into
Dentistas. Useful later as a **national sanity-check benchmark** once município‑level rates exist —
**not** usable for Q5 (that needs an EU/OECD comparison, which this file does not contain — see §5).


### 3.5 `Healthcare_perhabitant.xlsx` — ⚠️ empty export


```python
try:
    _df, _fields = load_pordata_quadro(RAW_DIR / "Healthcare_perhabitant.xlsx")
    print("Loaded OK:", _fields, len(_df), "rows")
except ValueError as e:
    print("PARSER RAISED (as expected):", e)

raw_empty = pd.read_excel(RAW_DIR / "Healthcare_perhabitant.xlsx", sheet_name="Quadro", header=None)
print("\nsheet shape:", raw_empty.shape)
print("Only non-empty content in the whole sheet is the symbol/legend key — no actual data table:")
raw_empty.iloc[13:19, 0:6]
```

**Finding: this file is empty.** It contains only PORDATA's footnote-symbol legend, no data rows at all
— the indicator was evidently not populated at export time. It contributes nothing today. **Action:**
re-export this indicator from PORDATA (or drop the file — `pharmacies_perthousand.xlsx` already covers

## 4. Join-key consistency across the município-level files

Before merging, confirm the three município-level files use identical, exactly-matching territory names
(no accents/whitespace drift that would silently drop rows on merge).


```python
names_density = set(density_df["Territorio"])
names_pharm   = set(pharm_df["Territorio"])
names_prof    = set(prof_df["Territorio"])

print("308 municípios in each file:",
      len(names_density) == 308, len(names_pharm) == 308, len(names_prof) == 308)
print("Exactly the same set of names across all three files:",
      names_density == names_pharm == names_prof)
print("Any leading/trailing whitespace in names:",
      any(n != n.strip() for n in names_density))
```

Clean result: all three files carry the **same 308 municípios** (278 mainland + 19 Açores + 11 Madeira —
matches Portugal's official município count), spelled identically, in the same order, with **zero missing
values**. These three merge safely on `Territorio` with an inner join and no row loss.


## 5. Data quality summary — is the raw data good enough?

| # | File | Level | Year(s) | Rows | Missing | Verdict |
|---|------|-------|---------|------|---------|---------|
| 1 | `Distribution_population.xlsx` | Município | 2025 only | 308/308 | 0 | ✅ Clean, merge-ready |
| 2 | `pharmacies_municipality.xlsx` | Município | 2025 only | 308/308 | 0 | ✅ Clean, merge-ready |
| 3 | `professionals_municipality.xlsx` | Município | 2025 only | 308/308 | 0 | ✅ Clean, merge-ready (Farmacêuticos under-counts nationals by ~2.8%, see §3.3) |
| 4 | `pharmacies_perthousand.xlsx` | Portugal (national) | 2025 only | 1 row | 0 | ⚠️ Usable only as a national sanity-check, not for município-level analysis |
| 5 | `Healthcare_perhabitant.xlsx` | — | — | 0 | — | ❌ Empty export, unusable as-is |

**Bottom line:** the three município-level files are genuinely good — complete, consistent, reconcilable
against official totals. That covers **raw counts and density**. But three things the business questions
need are **not in `Data/Raw` at all**, and no amount of cleaning recovers them:

1. **🚨 Município-level resident population (absolute counts).** `Distribution_population.xlsx` gives
   **density** (hab/km²), not population. Density ≠ population without also knowing each município's
   area — and area isn't in this file either. **This blocks Q1 as specified** ("doctors ... per 1,000
   residents in each município"): raw provider *counts* can be computed today, but a true **per-1,000-
   residents rate** cannot, because there is no population denominator yet. A large, sparsely-populated
   município (e.g. Odemira, one of the largest by area) and a small, dense one can share similar density
   figures while having very different populations — density is not a safe stand-in.
   → **Fetch PORDATA's "População residente" (or INE's annual resident-population estimates) by
   município, 2025**, same export format as the other three files. This single addition unblocks Q1, Q2
   and Q3 as designed.
2. **Only one reference year (2025) in every file.** Q4 asks whether the best-/worst-served gap has
   widened or narrowed over time. PORDATA's export UI defaults to the latest year unless "todos os anos"
   (all years) is selected at download time. → **Re-export each file with the full available year range.**
3. **No EU/OECD benchmark.** Q5 needs professionals-per-100,000 for the EU/OECD, not just Portugal.
   `pharmacies_perthousand.xlsx` gives Portugal's own 2025 national rate (useful as one side of that
   comparison) but the other side is missing. → **Eurostat (`hlth_rs_prsrg`) or OECD Health Statistics.**
4. **Q6's INFARMED dataset was already flagged unavailable** — confirmed here too: nothing INFARMED-sourced
   is present in `Data/Raw`. Q6 stays a stretch goal pending that source (or a substitute open dataset of
   pharmacy/mobile-unit siting).

None of this blocks starting — sections 6–9 below clean and explore everything that **is** usable today,
and are structured so adding the population file later is a one-line change (§9).


## 6. Build the clean, merged município-level dataset


```python
clean = (
    density_df
    .merge(pharm_df, on="Territorio", how="inner")
    .merge(prof_df, on="Territorio", how="inner")
    .rename(columns={
        "Territorio": "municipio",
        "Médicos": "medicos",
        "Enfermeiros": "enfermeiros",
        "Farmacêuticos": "farmaceuticos",
        "Dentistas": "dentistas",
    })
)
clean["total_profissionais_saude"] = clean[["medicos", "enfermeiros", "farmaceuticos", "dentistas"]].sum(axis=1)

assert len(clean) == 308, f"expected 308 municípios after merge, got {len(clean)}"
assert clean.isna().sum().sum() == 0, "unexpected missing values after merge"

print(clean.shape)
clean.head()
```

```python
out_path = CLEAN_DIR / "healthcare_access_municipality_2025.csv"
clean.to_csv(out_path, index=False)
print("saved:", out_path.resolve())
```

## 8. Distributions

All five variables are strongly **right-skewed** — expected for município-level counts, since a handful
of large urban municípios (Lisboa, Porto, Sintra, ...) dominate raw totals regardless of access quality.
This is exactly why Q1's per‑capita *rate* (once population is added) matters more than raw counts for the
"desert" story — raw counts alone would just re-tell "cities have more people."


```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
cols = ["densidade_hab_km2", "farmacias", "medicos", "enfermeiros", "farmaceuticos", "dentistas"]
for ax, col in zip(axes.flat, cols):
    sns.histplot(clean[col], bins=30, ax=ax, color="#3b6fa0")
    ax.set_title(col)
    ax.set_xlabel("")
plt.tight_layout()
plt.savefig(CLEAN_DIR / "../../Images/eda_distributions.png", dpi=110) if (CLEAN_DIR / "../../Images").exists() else None
plt.show()
```

## 9. Ranking municípios by density (population-count caveat applies — see §5.1)

These rankings use **density**, the one population-related field currently available. Once absolute
population is added, re-run this section with the true per-1,000-residents rate — the code is written so
that swap is a single column change (`RATE_BASIS` below).


```python
RATE_BASIS = "densidade_hab_km2"  # swap to a population column once §5.1's gap is filled

top15 = clean.nlargest(15, RATE_BASIS)[["municipio", RATE_BASIS]]
bottom15 = clean.nsmallest(15, RATE_BASIS)[["municipio", RATE_BASIS]]

fig, axes = plt.subplots(1, 2, figsize=(13, 6), sharex=False)
sns.barplot(data=top15, y="municipio", x=RATE_BASIS, ax=axes[0], color="#2a6f4b")
axes[0].set_title("Top 15 by population density")
sns.barplot(data=bottom15, y="municipio", x=RATE_BASIS, ax=axes[1], color="#b03a2e")
axes[1].set_title("Bottom 15 by population density")
plt.tight_layout()
plt.show()
```

## 10. Correlation structure

How raw provider counts relate to each other and to density. (Interpret cautiously per §5.1 — this is
counts and density, not access rates yet.)


```python
corr_cols = ["densidade_hab_km2", "farmacias", "medicos", "enfermeiros", "farmaceuticos", "dentistas"]
corr = clean[corr_cols].corr(method="spearman")

fig, ax = plt.subplots(figsize=(7, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1, ax=ax)
ax.set_title("Spearman correlation (municipality-level, n=308)")
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import spearmanr

print("Density vs each provider count (Spearman rank correlation, n=308):\n")
for col in ["farmacias", "medicos", "enfermeiros", "farmaceuticos", "dentistas"]:
    rho, p = spearmanr(clean["densidade_hab_km2"], clean[col])
    print(f"  {col:15s}  rho={rho:+.3f}   p={p:.2e}")
```

All five provider variables correlate strongly and significantly with density — but this is expected and
**does not yet answer Q2**. Q2 asks whether density predicts *access* (a rate) — raw counts scale with
population almost by construction, so a strong count↔density correlation is not evidence of an access
relationship. This correlation becomes meaningful for Q2 only after converting counts to per-capita rates
(§5.1).


## 11. Interior vs. coastal — a first (rough) cut

Using density terciles as a rough coastal/interior proxy (PORDATA's own território groupings — Alto Minho,
Douro, Beiras e Serra da Estrela, etc. — would be a more defensible interior/coastal split for the final
deliverable; that requires joining in each município's NUTS III region, which none of the five raw files
provide directly and is worth sourcing from INE's official município→NUTS III crosswalk).


```python
clean["density_tercile"] = pd.qcut(clean["densidade_hab_km2"], 3, labels=["low (interior-like)", "mid", "high (coastal-like)"])
summary = clean.groupby("density_tercile", observed=True)[["farmacias", "medicos", "enfermeiros", "farmaceuticos", "dentistas"]].mean().round(1)
summary
```
