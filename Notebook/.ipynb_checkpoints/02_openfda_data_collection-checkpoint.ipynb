## API Request

```python
url = (
    f"https://api.fda.gov/drug/event.json?"
    f"api_key={API_KEY}&limit=100"
)
```

```python
response = requests.get(url)

response.status_code
```

## Convert To DataFrame

```python
data = response.json()
```

```python
# Explore
data.keys()
```

```python
# Create dataframe

events_df = pd.json_normalize(
    data["results"]
)
```

```python
# inspection

events_df.head()
```

```python
# Save Data

events_df.to_csv(
    "../data/raw/openfda_events.csv",
    index=False
)
```
