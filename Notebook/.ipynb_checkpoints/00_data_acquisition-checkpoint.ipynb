# Data Acquisition — Healthcare Prescription Safety & Patient Risk Intelligence Platform

**Goal of this notebook:** pull real data from the four sources you shortlisted (MIMIC-IV, CMS Open Data, WHO API, OpenFDA), check what each one actually gives us against the six business questions, and land raw files in `Data/Raw/` for the cleaning/EDA notebook to pick up next.

**Business problem, restated as data questions:**

| # | Business need | What kind of data answers it |
|---|---|---|
| 1 | High-risk patients | Demographics + diagnoses + admission severity (MIMIC-IV) |
| 2 | Potential drug interactions | Drug labeling / interaction text, adverse events (OpenFDA) |
| 3 | Medication adherence issues | Refill/claims patterns — **none of the 4 sources below cover this directly**, see Part 5 |
| 4 | Readmission risks | Admission history (MIMIC-IV), hospital-level readmission rates (CMS) |
| 5 | Prescription fraud patterns | Prescriber billing volumes/outliers (CMS Part D Prescribers) |
| 6 | Chronic disease management | Diagnoses (MIMIC-IV), population benchmarks (WHO), disease indicators (CDC, Part 5) |

**How to run this:** these are live API calls, so run this notebook in your own local Jupyter/VS Code with normal internet access (not inside a locked-down sandbox). Everything here uses only `requests`, `pandas`, `pathlib`, `zipfile` — all included with a standard Anaconda/pip install. No paid accounts or API keys are required for anything below (OpenFDA works better with a free key — see Part 1.4 — but works without one too).


## 0. Setup

Creates one subfolder per source under `Data/Raw/` so raw pulls stay organized and don't get mixed with your cleaned data.


````python
import requests
import pandas as pd
import numpy as np
import json
import time
import zipfile
import io
from pathlib import Path

RAW_DIR = Path("../Data/Raw")
for sub in ["openfda", "who", "cms", "mimic_iv_demo"]:
    (RAW_DIR / sub).mkdir(parents=True, exist_ok=True)

pd.set_option("display.max_columns", 25)
pd.set_option("display.width", 140)

def get_json(url, params=None, timeout=30, retries=3, backoff=2):
    """GET a URL and return parsed JSON, with basic retry on transient failures.
    Raises for HTTP errors so problems surface immediately instead of silently
    returning nothing (important when you're new to APIs and debugging).
    """
    last_err = None
    for attempt in range(1, retries + 1):
        try:
            resp = requests.get(url, params=params, timeout=timeout)
            resp.raise_for_status()
            return resp.json()
        except requests.exceptions.RequestException as e:
            last_err = e
            print(f"  attempt {attempt}/{retries} failed: {e}")
            time.sleep(backoff * attempt)
    raise last_err

print("Raw data folders ready under:", RAW_DIR.resolve())
````

## Part 1 — OpenFDA API (`api.fda.gov`)

**Feasibility: ★★★★★ — best direct fit, start here.** Fully public, JSON REST API, no signup required to try it, optional free API key for higher rate limits. This is the one source that speaks directly to business question #2 (drug interactions) and contributes to #5 (fraud/quality signals via recalls).

Three endpoints we'll use:
- `drug/event.json` — adverse event reports (who took what, what reaction happened)
- `drug/enforcement.json` — recalls
- `drug/label.json` — official drug labels, including a `drug_interactions` free-text field

Docs: https://open.fda.gov/apis/drug/event/how-to-use-the-endpoint/


### 1.1 Adverse events — which drugs show up most in serious-outcome reports

````python
def fetch_openfda(endpoint, search=None, limit=100, max_records=500, api_key=None):
    """Page through an openFDA endpoint (event / enforcement / label) and return a list of records.
    openFDA caps `limit` at 1000 per call and total skip+limit at 25000 without a key
    (deep paging needs an API key from https://open.fda.gov/apis/authentication/ — free, instant).
    """
    base = f"https://api.fda.gov/drug/{endpoint}.json"
    records, skip = [], 0
    while len(records) < max_records:
        params = {"limit": min(limit, max_records - len(records)), "skip": skip}
        if search:
            params["search"] = search
        if api_key:
            params["api_key"] = api_key
        try:
            data = get_json(base, params=params)
        except requests.exceptions.HTTPError as e:
            if e.response is not None and e.response.status_code == 404:
                break  # no more results
            raise
        batch = data.get("results", [])
        if not batch:
            break
        records.extend(batch)
        skip += len(batch)
        time.sleep(0.25)  # be polite to the free, unauthenticated tier
    return records

# Example: serious adverse events reported for a high-interaction-risk drug (warfarin)
events = fetch_openfda("event", search='patient.drug.medicinalproduct:"warfarin"', max_records=200)
print(f"pulled {len(events)} adverse event reports")

events_df = pd.json_normalize(events)
events_df.to_json(RAW_DIR / "openfda" / "adverse_events_warfarin.json", orient="records")
events_df.head()
````

### 1.2 Drug recalls (enforcement reports) — quality/fraud-adjacent signal

````python
recalls = fetch_openfda("enforcement", search='status:"Ongoing"', max_records=200)
print(f"pulled {len(recalls)} ongoing recall records")

recalls_df = pd.json_normalize(recalls)
recalls_df.to_csv(RAW_DIR / "openfda" / "recalls_ongoing.csv", index=False)
recalls_df[["product_description", "reason_for_recall", "classification", "recall_initiation_date"]].head()
````

### 1.3 Drug labels — the actual interaction data

`drug_interactions` and `boxed_warning` are free-text fields written by the manufacturer/FDA — this is the closest thing to a real drug-interaction dataset that's fully open with no license. For a handful of named drugs this is easy to pull directly; for checking interactions between two *arbitrary* drugs at scale, see the DDInter alternative in Part 5.

````python
drugs_of_interest = ["warfarin", "metformin", "lisinopril", "atorvastatin", "omeprazole"]

label_rows = []
for drug in drugs_of_interest:
    labels = fetch_openfda("label", search=f'openfda.generic_name:"{drug}"', max_records=1)
    if not labels:
        print(f"no label found for {drug}")
        continue
    lbl = labels[0]
    label_rows.append({
        "drug": drug,
        "drug_interactions": " ".join(lbl.get("drug_interactions", [])),
        "boxed_warning": " ".join(lbl.get("boxed_warning", [])),
        "warnings": " ".join(lbl.get("warnings", []))[:500],
    })
    time.sleep(0.25)

labels_df = pd.DataFrame(label_rows)
labels_df.to_csv(RAW_DIR / "openfda" / "drug_labels_interactions.csv", index=False)
labels_df
````

### 1.4 Optional: get a free API key

Without a key you get **40 requests/min, 1,000/day** — plenty for exploration. Signing up (instant, no approval wait) at https://open.fda.gov/apis/authentication/ raises that to **240 requests/min, 120,000/day**, worth doing once you're pulling data for the whole project. Pass it as `api_key="YOUR_KEY"` to `fetch_openfda(...)` above.

## Part 2 — WHO Global Health Observatory (GHO) API

**Feasibility: ★★★★☆ — easy, but indirect.** Fully open OData API, no key, no signup. It won't give you patient- or prescription-level data — its role here is *benchmarking*: population-level chronic disease and mortality indicators to contextualize whatever patient cohort you build from MIMIC-IV (business question #6).

Base URL: `https://ghoapi.azureedge.net/api/`


````python
# Step 1: find indicator codes relevant to chronic disease / medicines
resp = requests.get(
    "https://ghoapi.azureedge.net/api/Indicator",
    params={"$filter": "contains(IndicatorName,'diabetes')"},
    timeout=30,
)
resp.raise_for_status()
indicators = pd.json_normalize(resp.json()["value"])
indicators[["IndicatorCode", "IndicatorName"]].head(15)
````

````python
# Step 2: pull data for a chosen indicator code (example: WHOSIS_000001 = life expectancy at birth,
# a standard benchmarking indicator; swap in a code from the diabetes search above once you've picked one)
INDICATOR_CODE = "WHOSIS_000001"

resp = requests.get(f"https://ghoapi.azureedge.net/api/{INDICATOR_CODE}", timeout=30)
resp.raise_for_status()
who_df = pd.json_normalize(resp.json()["value"])
who_df.to_csv(RAW_DIR / "who" / f"{INDICATOR_CODE}.csv", index=False)
print(f"{len(who_df)} rows")
who_df[["SpatialDim", "TimeDim", "NumericValue"]].head()
````

## Part 3 — CMS Open Data (`data.cms.gov`)

**Feasibility: ★★★★☆ — easy, directly relevant to #4 and #5.** No API key or registration needed. CMS consolidated its old Provider Data Catalog into one system, so the pattern below is: **search the catalog for a dataset by keyword → grab its identifier → query its rows.** This is more robust than hardcoding a dataset ID (those occasionally change), and it's a pattern you'll reuse for any open-data portal.

API guide: https://data.cms.gov/api-docs


````python
def find_cms_datasets(keyword, catalog_url="https://data.cms.gov/data.json"):
    """Search the CMS dataset catalog by keyword in the title. Returns (title, identifier) pairs."""
    catalog = get_json(catalog_url)
    hits = []
    for ds in catalog.get("dataset", []):
        title = ds.get("title", "")
        if keyword.lower() in title.lower():
            hits.append((title, ds.get("identifier")))
    return hits

readmission_hits = find_cms_datasets("Readmission")
fraud_hits = find_cms_datasets("Prescriber")

print("Readmission-related datasets:")
for title, ident in readmission_hits:
    print(" -", title, "|", ident)

print("\\nPrescriber-related datasets (relevant to fraud-pattern detection):")
for title, ident in fraud_hits:
    print(" -", title, "|", ident)
````

````python
# Once you've picked an identifier from the printed list above, pull its rows:
DATASET_UUID = readmission_hits[0][1] if readmission_hits else None

if DATASET_UUID:
    data_url = f"https://data.cms.gov/data-api/v1/dataset/{DATASET_UUID}/data"
    resp = requests.get(data_url, params={"size": 5000, "offset": 0}, timeout=60)
    resp.raise_for_status()
    cms_df = pd.DataFrame(resp.json())
    cms_df.to_csv(RAW_DIR / "cms" / "readmissions_reduction_program.csv", index=False)
    print(f"{len(cms_df)} rows saved")
    display(cms_df.head())
else:
    print("No match found in this catalog pull — the Hospital Readmissions Reduction Program dataset "
          "also lists its own API tab at https://data.cms.gov/provider-data/dataset/9n3s-kdb3 ; "
          "open that page in a browser, click the 'API' tab, and copy the identifier shown there.")
````

## Part 4 — MIMIC-IV

**Feasibility of the FULL database: ★☆☆☆☆ — not realistic for this project's timeline.** MIMIC-IV (the real, ~300k-admission database) requires:
1. A free PhysioNet account
2. Completing the CITI Program "Data or Specimens Only Research" training course (a multi-hour, multi-module course with a quiz)
3. Signing a Data Use Agreement and being approved as a *credentialed user* (manual review, can take days)

That's a real barrier for a bootcamp timeline — budget for it only if you have 1-2 weeks of slack and want ICU-level detail later.

**Feasibility of the MIMIC-IV Demo: ★★★★★ — use this instead.** Same schema (26 tables), 100 real de-identified patients, **zero credentialing** — openly licensed under Open Data Commons Open Database License v1.0. Good enough to build and demo your whole pipeline (admissions, diagnoses, ICU stays, demographics — business questions #1 and #4) before deciding whether the full database is worth the credentialing effort.

Source: https://physionet.org/content/mimic-iv-demo/2.2/


````python
MIMIC_ZIP_URL = "https://physionet.org/content/mimic-iv-demo/get-zip/2.2/"
mimic_dir = RAW_DIR / "mimic_iv_demo"

resp = requests.get(MIMIC_ZIP_URL, timeout=120)
resp.raise_for_status()

if resp.headers.get("Content-Type", "").startswith("application/zip") or resp.content[:2] == b"PK":
    with zipfile.ZipFile(io.BytesIO(resp.content)) as zf:
        zf.extractall(mimic_dir)
    print("Extracted MIMIC-IV demo to", mimic_dir.resolve())
else:
    print("Did not receive a zip file (got Content-Type:", resp.headers.get("Content-Type"), ")")
    print("This usually means PhysioNet served a login/consent page instead of the file.")
    print("Fallback: download manually in a browser from")
    print("  https://physionet.org/content/mimic-iv-demo/2.2/")
    print(f"and unzip into {mimic_dir.resolve()}")
````

````python
# Load a few key tables once extracted (paths follow physionet's hosp/ and icu/ folder layout)
import glob

def find_and_read(pattern):
    matches = glob.glob(str(mimic_dir / "**" / pattern), recursive=True)
    if not matches:
        print(f"not found yet: {pattern} (did the download/extract above succeed?)")
        return None
    return pd.read_csv(matches[0])

patients_df    = find_and_read("patients.csv*")
admissions_df  = find_and_read("admissions.csv*")
diagnoses_df   = find_and_read("diagnoses_icd.csv*")
icustays_df    = find_and_read("icustays.csv*")

for name, df in [("patients", patients_df), ("admissions", admissions_df),
                  ("diagnoses_icd", diagnoses_df), ("icustays", icustays_df)]:
    if df is not None:
        print(f"{name}: {df.shape}")
````

## Part 5 — Alternative & supplementary sources

The four sources above don't fully cover business question #3 (**medication adherence**) and only partly cover #5 (**fraud**) and #6 (**chronic disease benchmarking**). Here's what to add, roughly in order of how much they'd help relative to setup effort:

### 5.1 CMS Medicare Part D Prescribers (by Provider and Drug) — fills gap in #5 (fraud) and #3 (adherence proxy)
Same `data.cms.gov` catalog/API pattern as Part 3 — search for `"Part D Prescribers"` with `find_cms_datasets()`. Gives, per prescriber and drug: claim counts, days supplied, beneficiary counts. Unusually high claim volume relative to peers in the same specialty is the standard proxy for **prescription fraud pattern** detection, and days-supplied/refill-gap fields are a workable proxy for **adherence** since no source here has real pharmacy-refill data.

### 5.2 Synthea — synthetic patient generator (fills any remaining gap, zero privacy/licensing friction)
Open-source (MITRE), generates **fully synthetic but clinically realistic** patient records — including a `medications.csv` with start/stop dates you can turn into an adherence/gap metric, plus encounters, conditions, and readmission-shaped visit histories. No credentialing, no PHI risk, and you control cohort size/disease mix.
```bash
git clone https://github.com/synthetichealth/synthea.git
cd synthea
./run_synthea -p 1000   # generates ~1000 synthetic patients as CSV/FHIR
```
Best use here: generate a synthetic "medication adherence" cohort to cover business question #3, since none of your four primary sources has refill-level data.

### 5.3 DDInter 2.0 — structured drug-drug interactions (upgrade path from OpenFDA's free text)
A curated, downloadable database of drug pairs with interaction **severity levels** (major/moderate/minor) — more analysis-ready than OpenFDA's free-text `drug_interactions` field, since it's already structured for a lookup/scoring approach. Worth switching to this for #2 once you've prototyped with OpenFDA. https://ddinter2.scbdd.com/

### 5.4 CDC PLACES / Chronic Disease Indicators — sharper benchmarking than WHO for #6
If your patient cohort is US-based (which it will be, from MIMIC-IV), CDC's PLACES dataset gives **US county-level** chronic disease prevalence (diabetes, hypertension, COPD, etc.) — a more relevant benchmark than WHO's country-level global figures. Same open, no-key REST API pattern as CMS: https://data.cdc.gov/resource/cwsq-ngmh.json (Socrata `$limit`/`$where` query params).

### 5.5 Kaggle — fastest path to a clean, ready-made CSV if you're short on time
Not primary sources, but useful if any of the above stalls close to a deadline: search Kaggle for "Medicare Part D", "healthcare readmission", or "drug reviews" datasets. Requires a free Kaggle account + API token (`~/.kaggle/kaggle.json`), then:
```python
import kagglehub
path = kagglehub.dataset_download("<owner>/<dataset-slug>")
```

This combination is fully assembled with **zero paid accounts, zero approval-wait credentialing, and zero scraping** — every pull above is a documented, public REST API. That matters for a beginner project: it's reproducible by anyone reviewing your work, and you're not depending on a credentialing approval that might not arrive before your deadline.
